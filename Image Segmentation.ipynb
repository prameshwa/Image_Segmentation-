```python
import cv2
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
```

```python
print("OpenCV version:", cv2.__version__)
```

```python

from sklearn.cluster import KMeans
```

```python
print ("All libraries imported successfully!")
```

```python
Image = mpl.image.imread(r'Original\bat_ball2.jpg')
```

```python
plt.imshow(Image)
plt.axis('off')
```

```python
x = Image.reshape((-1,3))
```

```python
kmeans = KMeans(n_clusters=4, n_init=10)
```

```python
kmeans.fit(x)
```

```python
segmented_img = kmeans.cluster_centers_[kmeans.labels_]
segmented_img = segmented_img.reshape(Image.shape)
```

```python
plt.imshow(segmented_img / 255)
plt.axis('off')
```

```python
cv2.imwrite('image1.png', cv2.cvtColor(Image, cv2.COLOR_RGB2BGR))
cv2.imwrite('image2.png', cv2.cvtColor(segmented_img.astype('uint8'), cv2.COLOR_BGR2HSV))
```
